 Q1. What is Simple Linear Regression (SLR)? Explain its purpose?

 Simple Linear Regression (SLR) is a basic statistical method. It models the straight-line relationship between one independent input variable (\(x\)) and one dependent output variable (\(y\)). It finds a best-fit line to show how changes in the input cause changes in the output.

 # The Purpose of SLR

 * **Prediction:** Estimate future or unknown output values (\(y\)) based on new input values (\(x\)).  
 * **Relationship Analysis:** Measure if and how strongly two variables connect to each other.   
 * **Trend Evaluation:** See the direction and speed of change between data points.
 #How SLR Works
 **The Equation: **
* y is the predicted output
* x is the input value.
* b(0) is the intercept (where the line crosses the vertical axis).
* b(1) is the slope (how steep the line is and how much (y) changes when (x) grows).

**Best-Fit Line:** Uses the least-squares method to draw a line that cuts through data points with the smallest total error distance.   
 # Common Examples

 **Advertising vs Sales:** Using monthly ad spending (x)to guess future sales income (y).   
 **Experience vs Salary:** Using years at a job (x) predict an employee's pay rate (y).

 **Size vs Cost:** Using a house size in square feet (x) to forecast its market price (y)



Q2. What are the key assumptions of Simple Linear Regression?

Simple linear regression relies on four key assumptions: a linear relationship between variables, independence of errors, homoscedasticity (constant variance), and normality of residuals.
#Core Assumptions
1. **Linearity:** The relationship between the input
variable (X) and the output variable (Y) must form a straight-line pattern.   
2. **Independence:** The error or residual terms of one observation must not match or depend on the errors of another observation.   
3. **Homoscedasticity:** The spread or variance of the error terms must stay the same across all values of X.  
**Normality:** The residual errors should follow a normal distribution (bell-curve shape) around zero.

Q3. Write the mathematical equation for a simple linear regression model and explain each term?

The mathematical equation for a simple linear regression model is y=b(0)+b(1)X+e where (Y) is the dependent variable, (X) is the independent variable, b(0) is the y-intercept, b(1) is the slope coefficient, and (e) is the error term.
#Equation Terms
* (Y): The dependent variable or response that you want to predict.
* (X): The independent variable or predictor used to explain the changes in (Y).
* b(0): The y-intercept; the expected value of (Y) when (X) equals zero.
* b(1): The slope coefficient; the amount (Y) changes for every one-unit increase in (X).
* (e): The error term or residual; the difference between the actual value and the predicted value.

Q5. What is the method of least squares in linear regression?

The least squares method is a mathematical way to find the best-fit line for a set of data points. It works by making the total squared vertical distance—called residuals—between the real data points and the line as small as possible.
#How It Works
**Define the line:** Assume a linear model (y = mx + c) where (m) is the slope and (c) is the y-intercept.

 **errors:** Find the difference between each actual data value (y) and the predicted value (^y) on the line.

**Square the errors:** Square each difference so positive and negative errors do not cancel out.

**Minimize the sum:** Add up all the squared errors and use calculus to find the values of (m) and (c) that yield the absolute lowest total

Q6. What is Logistic Regression? How does it differ from Linear Regression?
Logistic regression is a data method used to predict a category, such as "yes" or "no". It differs from linear regression because linear regression predicts numbers (like a house price), while logistic regression predicts groups or chances using a curved line.
#What is Logistic Regression?
* It is a tool to find out the chance of an event happening (like a number from 0 to 1).
* It uses a special math trick called a sigmoid function to bend a straight line into an S-shape.
* It helps solve classification jobs, like telling if an email is spam or not.

#Key Differences
* **Type of Answer:** Linear regression gives a continuous number (e.g., $50,000 salary). Logistic regression gives a class or group (e.g., pass or fail).
 * **The Line on the Graph:** Linear regression draws a straight line. Logistic regression draws an S-shaped curve.
 * **Math Rules:** Linear regression uses least squares to cut down errors. Logistic regression uses maximum likelihood estimation.

Q7. Name and briefly describe three common evaluation metrics for regression models?

Three common evaluation metrics for regression models are Mean Absolute Error (MAE), which measures average absolute distance; Mean Squared Error (MSE), which averages the squared error values; and R-squared (R²), which shows the proportion of explained variance.
# Mean Absolute Error (MAE)
* Finds the absolute difference between each real value and predicted value.
* Takes the average of these values.
* Uses the exact same unit as the original data.
* Treats all errors the exact same way without extra weight.
# Mean Squared Error (MSE)
* Squares the difference between each real value and predicted value.
* Computes the average of those squared values.
* Punishes large mistakes much harder than small mistakes.
* Sensitive to extreme outlier points in the data.
# R-squared (R² Score)
* Shows how well the model fits the overall variance.
* Ranges from values near 0 up to 1 (perfect fit).
* Compares your model performance against a simple baseline mean.
* Tells you what percentage of data variation your features explain.

Q8.What is the purpose of the R-squared metric in regression analysis?
The purpose of(R\2) (the coefficient of determination) is to measure the goodness of fit of a regression model by showing the proportion of variance in the dependent variable explained by the independent variable(s). It ranges from 0 to 1, where 0 means the model explains nothing and 1 means a perfect fit.
#Core Functions of(R\2)
* **Goodness of fit:** Shows how close the data points are to the fitted regression line.
* **Variance explanation:** Tells the percentage of variation in the target outcome (Y) accounted for by the predictor(s) (X).
* **Model comparison:** Helps compare how well different models fit the exact same data set.
# Key Limits to Remember
* **No quality guarantee:** A high(R\2) does not mean a model is free of bias, and a low (R\2) can still belong to a useful model.
* **Adds up with predictors:** Adding more independent variables naturally makes (R\2) rise or stay the same, which can lead to over-fitting unless you use an adjusted metric.

Q9. Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.    
  (Include your Python code and output in the code box below.)

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Generate sample data: y = 2x + 1
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([3, 5, 7, 9, 11])

# Initialize and fit the simple linear regression model
model = LinearRegression()
model.fit(X, y)

# Extract and print the slope (coefficient) and intercept
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (m): {slope}")
print(f"Intercept (c): {intercept}")
```
